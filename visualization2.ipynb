Preprocess Data (Python)

```python
import pandas as pd
import json
import os

# Load & clean data (same as your current code)
articles = pd.read_excel("files/articles.xlsx", engine="openpyxl")
authors = pd.read_excel("files/authors.xlsx", engine="openpyxl")

articles.columns = articles.columns.str.strip().str.lower()
authors.columns = authors.columns.str.strip().str.lower()

merged_data = authors.merge(articles[['article id', 'title']], on="article id", how="left")

author_publications = merged_data["author name"].value_counts().reset_index()
author_publications.columns = ["author name", "publication count"]
author_publications["publication count"] = author_publications["publication count"].astype(int)

if "country" in authors.columns:
    author_publications = author_publications.merge(
        authors[['author name', 'country']],
        on="author name",
        how="left"
    )
else:
    author_publications["country"] = "Unknown"

# Export to JSON
output_json = "files/author_data.json"
author_publications.to_json(output_json, orient="records", indent=2)
print(f"Saved processed data to: {output_json}")
```
